```python
%%bash
cat > /tmp/render_example.py <<'PY'
import gymnasium as gym
import mani_skill.envs
import imageio.v3 as iio
import numpy as np

env = gym.make(
    "PickCube-v1",
    num_envs=1,
    robot_uids="panda",
    obs_mode="rgbd",
    control_mode="pd_ee_delta_pose",
)

try:
    obs, _ = env.reset(seed=0)

    rgb = obs["sensor_data"]["base_camera"]["rgb"][0]
    depth = obs["sensor_data"]["base_camera"]["depth"][0]

    if hasattr(rgb, "cpu"):
        rgb = rgb.cpu().numpy()

    if hasattr(depth, "cpu"):
        depth = depth.cpu().numpy()

    depth = depth[..., 0]

    iio.imwrite("/content/example_rgb.png", rgb)
    np.save("/content/example_depth.npy", depth)

    print("RGB saved:", rgb.shape)
    print("Depth saved:", depth.shape)
finally:
    env.close()
PY

conda run -n maniskill python /tmp/render_example.py
```

#**one transition**

```python
%%bash
cat > /tmp/quickstart_transition.py <<'PY'
import gymnasium as gym
import mani_skill.envs
import imageio.v3 as iio
import numpy as np

env = gym.make(
    "PickCube-v1",
    num_envs=1,
    robot_uids="panda",
    obs_mode="rgbd",
    control_mode="pd_ee_delta_pose",
)

try:
    obs_t, _ = env.reset(seed=0)

    action_t = env.action_space.sample()

    obs_tp1, reward, terminated, truncated, info = env.step(action_t)

    rgb_t = obs_t["sensor_data"]["base_camera"]["rgb"][0]
    depth_t = obs_t["sensor_data"]["base_camera"]["depth"][0]

    rgb_tp1 = obs_tp1["sensor_data"]["base_camera"]["rgb"][0]
    depth_tp1 = obs_tp1["sensor_data"]["base_camera"]["depth"][0]

    def to_numpy(x):
        return x.cpu().numpy() if hasattr(x, "cpu") else np.asarray(x)

    rgb_t = to_numpy(rgb_t)
    depth_t = to_numpy(depth_t)[..., 0]

    rgb_tp1 = to_numpy(rgb_tp1)
    depth_tp1 = to_numpy(depth_tp1)[..., 0]

    np.save("/content/depth_t.npy", depth_t)
    np.save("/content/depth_tp1.npy", depth_tp1)

    iio.imwrite("/content/rgb_t.png", rgb_t)
    iio.imwrite("/content/rgb_tp1.png", rgb_tp1)

    np.save("/content/action.npy", to_numpy(action_t))

    print("TRANSITION CREATED")
    print("RGB before:", rgb_t.shape)
    print("Depth before:", depth_t.shape)
    print("Action:", to_numpy(action_t))
    print("RGB after:", rgb_tp1.shape)
    print("Depth after:", depth_tp1.shape)
    print("Reward:", reward)
    print("Terminated:", terminated)
    print("Truncated:", truncated)
finally:
    env.close()
PY

conda run -n maniskill python /tmp/quickstart_transition.py
```

```python
%%bash
cat > /tmp/render_visible_arm.py <<'PY'
import gymnasium as gym
import mani_skill.envs
from mani_skill.utils import sapien_utils
import imageio.v3 as iio
import numpy as np

env = gym.make(
    "PickCube-v1",
    num_envs=1,
    robot_uids="panda",
    obs_mode="rgbd",
    control_mode="pd_ee_delta_pose",
    sensor_configs=dict(
        base_camera=dict(
            pose=sapien_utils.look_at(
                eye=[0.55, 0.75, 0.75],
                target=[0.0, 0.0, 0.18],
            ),
            width=224,
            height=224,
            fov=1.1,
        )
    ),
)

try:
    obs, _ = env.reset(seed=0)

    def copy_array(value):
        if hasattr(value, "detach"):
            return value.detach().cpu().numpy().copy()
        return np.asarray(value).copy()

    # Copy the initial frame before stepping.
    rgb_before = copy_array(
        obs["sensor_data"]["base_camera"]["rgb"][0]
    )

    depth_before = copy_array(
        obs["sensor_data"]["base_camera"]["depth"][0]
    )[..., 0]

    # Create a valid action using the environment's actual action shape.
    action = np.zeros(env.action_space.shape, dtype=np.float32)

    # Move the end-effector upward and sideways.
    action[0] = 0.7
    action[2] = 0.7

    for _ in range(20):
        obs, reward, terminated, truncated, info = env.step(action)

    # Copy the final frame after stepping.
    rgb_after = copy_array(
        obs["sensor_data"]["base_camera"]["rgb"][0]
    )

    depth_after = copy_array(
        obs["sensor_data"]["base_camera"]["depth"][0]
    )[..., 0]

    iio.imwrite("/content/rgb_before.png", rgb_before)
    iio.imwrite("/content/rgb_after.png", rgb_after)

    np.save("/content/depth_before.npy", depth_before)
    np.save("/content/depth_after.npy", depth_after)

    diff = np.abs(
        rgb_after.astype(np.int16) - rgb_before.astype(np.int16)
    )

    print("Action:", action)
    print("RGB shape:", rgb_before.shape)
    print("Maximum pixel difference:", diff.max())
    print("Mean pixel difference:", diff.mean())
    print("Saved visible movement example")

finally:
    env.close()
PY

conda run -n maniskill python /tmp/render_visible_arm.py
```

#**closed-loop waypoint controller**

```python
%%bash
cat > /tmp/grasp_cube_high.py <<'PY'
import gymnasium as gym
import mani_skill.envs
from mani_skill.utils import sapien_utils
import imageio.v3 as iio
import numpy as np

env = gym.make(
    "PickCube-v1",
    num_envs=1,
    robot_uids="panda",
    obs_mode="rgbd",
    control_mode="pd_ee_delta_pose",
    sensor_configs=dict(
        base_camera=dict(
            pose=sapien_utils.look_at(
                eye=[0.55, 0.75, 0.75],
                target=[0.0, 0.0, 0.18],
            ),
            width=224,
            height=224,
            fov=1.1,
        )
    ),
)

try:
    obs, _ = env.reset(seed=0)

    def to_numpy(value):
        if hasattr(value, "detach"):
            return value.detach().cpu().numpy().copy()
        if hasattr(value, "cpu"):
            return value.cpu().numpy().copy()
        return np.asarray(value).copy()

    def save_rgb(observation, filename):
        rgb = to_numpy(
            observation["sensor_data"]["base_camera"]["rgb"][0]
        )
        iio.imwrite(f"/content/{filename}.png", rgb)

    def get_tcp_position(observation):
        return to_numpy(
            observation["extra"]["tcp_pose"][0, :3]
        )

    def get_cube_position():
        return to_numpy(env.unwrapped.cube.pose.p[0])

    def move_to(observation, target_position, gripper_value, steps, filename):
        for _ in range(steps):
            tcp_position = get_tcp_position(observation)
            delta = target_position - tcp_position

            action = np.zeros(
                env.action_space.shape,
                dtype=np.float32,
            )

            action[:3] = np.clip(delta / 0.1, -1.0, 1.0)
            action[3:6] = 0.0
            action[6] = gripper_value

            observation, reward, terminated, truncated, info = env.step(action)

        save_rgb(observation, filename)
        return observation

    save_rgb(obs, "grasp_00_start")

    cube_position = get_cube_position()
    print("Cube position:", cube_position)

    # Move above the cube.
    above_cube = cube_position + np.array([0.0, 0.0, 0.16])
    obs = move_to(
        obs,
        above_cube,
        gripper_value=1.0,
        steps=15,
        filename="grasp_01_above",
    )

    # Move down around the cube.
    grasp_position = cube_position + np.array([0.0, 0.0, 0.025])
    obs = move_to(
        obs,
        grasp_position,
        gripper_value=1.0,
        steps=15,
        filename="grasp_02_near_cube",
    )

    # Close gripper.
    for _ in range(20):
        action = np.zeros(env.action_space.shape, dtype=np.float32)
        action[6] = -1.0
        obs, reward, terminated, truncated, info = env.step(action)

    save_rgb(obs, "grasp_03_gripper_closed")

    is_grasped = bool(
        to_numpy(obs["extra"]["is_grasped"])[0]
    )

    print("Is grasped:", is_grasped)

    if is_grasped:
        # First lift.
        lift_position_1 = cube_position + np.array([0.0, 0.0, 0.30])
        obs = move_to(
            obs,
            lift_position_1,
            gripper_value=-1.0,
            steps=30,
            filename="grasp_04_lifted",
        )

        # Much higher lift.
        lift_position_2 = cube_position + np.array([0.0, 0.0, 0.50])
        obs = move_to(
            obs,
            lift_position_2,
            gripper_value=-1.0,
            steps=45,
            filename="grasp_05_lifted_high",
        )

        # Hold the cube in the high position.
        for _ in range(20):
            action = np.zeros(env.action_space.shape, dtype=np.float32)
            action[6] = -1.0
            obs, reward, terminated, truncated, info = env.step(action)

        save_rgb(obs, "grasp_06_held_high")
        print("SUCCESS: cube lifted and held high.")

    else:
        save_rgb(obs, "grasp_04_failed")
        print("FAILED: cube was not grasped.")

finally:
    env.close()
PY

conda run -n maniskill python /tmp/grasp_cube_high.py
```

```python
from PIL import Image
import matplotlib.pyplot as plt
from pathlib import Path

names = [
    "grasp_00_start",
    "grasp_01_above",
    "grasp_02_near_cube",
    "grasp_03_gripper_closed",
    "grasp_04_lifted",
    "grasp_05_lifted_high",
    "grasp_06_held_high",
]

names = [
    name for name in names
    if Path(f"/content/{name}.png").exists()
]

fig, axes = plt.subplots(1, len(names), figsize=(24, 5))

for axis, name in zip(axes, names):
    axis.imshow(Image.open(f"/content/{name}.png"))
    axis.set_title(name.replace("grasp_", "").replace("_", " "))
    axis.axis("off")

plt.tight_layout()
plt.show()
```
